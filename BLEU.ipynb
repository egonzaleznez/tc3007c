```python
df = pd.read_excel("Evaluación de la calidad de respuestas de un chatbot(1-35).xlsx",)
```

```python
responses = open("respuestas1.txt",'+r')
response_list = responses.readlines()
```

```python
clean_response = []
chat_responses = []
user_question = []
```

```python
for element in response_list:
    res = ''.join([char for char in element if char.isalnum() or char==' ' or char =='.' or char ==',' or char =='?' or char =='!'])
    if res !='':
        clean_response.append(res)

for i in range(len(clean_response)):
    
    if i%2==0:
        user_question.append(clean_response[i])
    else:
        chat_responses.append(clean_response[i])
    
```

```python
import spacy

# Cargar modelo de español
nlp = spacy.load("es_core_news_sm")

# Tokenizar con spaCy
def tokenize_with_spacy(text):
    doc = nlp(text.lower())
    return [token.text for token in doc]

tokenized_references = [tokenize_with_spacy(ref) for ref in expert_response]
tokenized_candidate = tokenize_with_spacy(chat_responses[0])
#tokenized_references = [tokenize_with_spacy(ref) for ref in ["El gato negro corre en la avenida, es muy rapido"]]
#tokenized_candidate = tokenize_with_spacy("El  negro corre en la es muy rapido")

# Cálculo de BLEU
score = sentence_bleu(tokenized_references, tokenized_candidate)
print(f"BLEU score (con spaCy): {score:.4f}")
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Cargar modelo multilingüe (soporta español)
model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

# Frases a comparar
frase1 = "El consumo de alcohol puede tener efectos negativos en desarrollo del bebé, consulte a su médico para más especificaciones. "
frase2 =chat_responses[0]

# Obtener embeddings
embedding1 = model.encode([frase1])
embedding2 = model.encode([frase2])

# Calcular similitud de coseno
similitud = cosine_similarity(embedding1, embedding2)[0][0]
print(f"Similitud semántica: {similitud:.4f}")
```
